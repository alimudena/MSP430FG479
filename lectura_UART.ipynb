```python
file_name = "Untitled_1.stc"

# Leer el contenido como texto
with open(file_name, "r", encoding="utf-8") as file:
    content = file.read().strip()  # Eliminamos espacios y saltos de línea innecesarios

# Procesar contenido hexadecimal
try:
    # Filtramos para quedarnos solo con caracteres hexadecimales
    hex_data = ''.join(filter(str.isalnum, content))  # Solo dejamos caracteres alfanuméricos

    # Dividimos en bytes (2 caracteres por byte)
    bytes_data = [hex_data[i:i+2] for i in range(0, len(hex_data), 2)]

    # Agrupamos en bloques de 16 bits (2 bytes)
    blocks_16bit = [bytes_data[i] + bytes_data[i+1] for i in range(0, len(bytes_data)-1, 2)]

    print("Contenido procesado en bloques de 16 bits:")
    print(blocks_16bit)
except ValueError as e:
    print("El archivo no contiene datos válidos en hexadecimal.")
```

```python
# Abrir el archivo en modo binario
with open('test_sine.txt', 'rb') as file:
    # Leer el contenido del archivo
    byte_data = file.read()

# Mostrar el contenido en formato hexadecimal
hex_representation = byte_data.hex()

# Mostrar los primeros 100 caracteres de la representación hexadecimal (si el archivo es grande)
print(hex_representation[:100])
```

```python
hex_representation
```

```python
# Agrupar los caracteres hexadecimales en bloques de 4 caracteres
grouped_hex = [hex_representation[i:i+4] for i in range(0, len(hex_representation), 4)]

# Mostrar el resultado
print(grouped_hex[:10])  # Muestra los primeros 10 grupos
```

```python
# Convertir cada bloque hexadecimal en binario
grouped_bin = [bin(int(group, 16))[2:].zfill(16) for group in grouped_hex]

# Mostrar los primeros 10 valores binarios
print(grouped_bin[:10])
```

```python
# Convertir cada bloque hexadecimal a un número sin signo (uint16_t)
grouped_uint16 = [int(group, 16) & 0xFFFF for group in grouped_hex]

# Mostrar los primeros 10 valores uint16_t
print(grouped_uint16[:10])
```

```python
import matplotlib.pyplot as plt

# Suponiendo que 'grouped_uint16' ya contiene los valores que deseas graficar
# Ejemplo de 'grouped_uint16'
# grouped_uint16 = [6699, 15405, 94, ...]
# Crear un gráfico de líneas
plt.plot(grouped_uint16, linestyle='-', color='b')

# Etiquetas
plt.xlabel('Índice')
plt.ylabel('Valor (uint16_t)')
plt.title('Valores hexadecimales representados como uint16_t')
plt.ylim(0, 65536)  # Limitar el eje Y entre 0 y 65536 (como el rango de uint16_t)
plt.xlim(0, 1000)
# Mostrar la gráfica
plt.show()
```
